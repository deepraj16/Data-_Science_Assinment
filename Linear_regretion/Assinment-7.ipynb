```python
# Outlier  remove 
def range_of_limit(col): 
    list=[]
    mean=df[col].mean()
    std = df[col].std()
    upper_limte = mean + 3*std 
    lower_limte =mean - 3*std
    return upper_limte,lower_limte 
```

```python
def capping(col ) : 
    upper_limte , lower_limte = range_of_limit(col)
    r = np.where(df[col]>upper_limte 
                          , upper_limte , 
                         np.where(
                             df[col]<lower_limte , lower_limte,df[col]
                         ))
    return r
```
